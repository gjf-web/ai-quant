```python
before = id(Y)
Y = Y + X
id(Y) == before
#id() gives us the exact address of the referenced object in memory
```

```python
#to avoid data move, we use in-place operation to save memory!
Z = torch.zeros_like(Y) #_like: have the same shape as Y
print("id(Z):", id(Z))
Z[:] = X + Y #[:]: use the original address, keep address stay still 
print('id(Z):',id(Z))
```
